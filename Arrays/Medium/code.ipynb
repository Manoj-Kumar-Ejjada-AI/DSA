```python
def nextPermutation(self, nums):
    # Set index
    index = -1

    # find decreasing point
    for _ in range(len(nums) - 2, -1, -1):
        # If smaller found
        if nums[_] < nums[_ + 1]:
            index = _
            break

    # if no decreasing index
    if index == -1:
        # reverse complete list
        nums.reverse()
        return

    # find just greater element
    for i in range(len(nums) - 1, index, -1):
        if nums[_] > nums[index]:
            # swap them
            nums[_], nums[index] = nums[index], nums[_]
            break

    # reverse part of array after index
    nums[index + 1:] = reversed(nums[index + 1:])
```
