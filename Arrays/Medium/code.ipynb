```python
def nextPermutation(arr):
    index = -1

    for i in range(len(arr)-1,0,-1):
        if arr[i]>arr[i-1]:
            index = i-1
    if index == -1:
        arr.reverse()
        return
    for i in range(len(arr)-1,index,-1):
        if arr[i]>arr[index]:
            arr[i],arr[index] =arr[index],arr[i]
            break
    arr[index+1:] = reversed(arr[index+1:])
```
