# Functional Recursion

## Print sum of first `N` numbers

```python
def _sum(n):
    if n == 1:
        return 1
    return n + _sum(n-1)
print(_sum(3))
```

#### Time Comlexity O(N)
#### Space Complexity O(N)

## Factorial

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)
print(factorial(5))
```

#### Time Comlexity O(N)
#### Space Complexity O(N)
