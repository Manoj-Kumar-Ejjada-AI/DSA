# Functional Recursion

## Print sum of first `N` numbers

````python
def _sum(n):
    if n == 1:
        return 1
    return n + _sum(n-1)
print(_sum(3))
````

#### Time Comlexity O(N)
#### Space Complexity O(N)

## Factorial

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)
print(factorial(5))
````

#### Time Comlexity O(N)
#### Space Complexity O(N)

## Reverse an Array

````python
def reverse_array(arr, low, high):
    if high<=low:
        return
    arr[low],arr[high] = arr[high], arr[low]
    reverse_array(arr,low+1,high-1)
````

````python
l = [1,2,3,4,5]
reverse_array(l,0,4)
l
````

#### Time Complexity O(N)
#### Space Complexity O(N)

### Why time complexity is not O(logn)
It's a common point of confusion. The time complexity is **not** (O(\log n)). Here's why.

### What happens in each recursive call?

```python
reverse_array(arr, low+1, high-1)
```

Suppose the array has `n` elements.

Each recursive call reduces the problem size by **2 elements**, not by half.

For example, if `n = 8`:

```
reverse(0, 7)  # 8 elements
reverse(1, 6)  # 6 elements
reverse(2, 5)  # 4 elements
reverse(3, 4)  # 2 elements
reverse(4, 3)  # stop
```

The sizes are:

```
8 → 6 → 4 → 2 → 0
```

You're subtracting **2** each time.

---

### Recurrence Relation

The recurrence is:

[
T(n) = T(n-2) + O(1)
]

Expanding:

```
T(n)
= T(n-2) + c
= T(n-4) + 2c
= T(n-6) + 3c
...
```

After approximately `n/2` calls:

```
T(n) = c * (n/2) = O(n)
```

---

### When do we get (O(\log n))?

You get logarithmic time only when the problem size is divided by a constant factor each step.

Examples:

```python
binary_search(arr, low, high)
```

Each call searches only half the array:

```
n → n/2 → n/4 → n/8 → ...
```

Recurrence:

[
T(n) = T(n/2) + O(1)
]

which solves to:

[
O(\log n)
]

---

### Comparison

| Problem size after each call | Recurrence         | Time Complexity |
| ---------------------------- | ------------------ | --------------- |
| `n → n-1`                    | `T(n)=T(n-1)+O(1)` | **O(n)**        |
| `n → n-2`                    | `T(n)=T(n-2)+O(1)` | **O(n)**        |
| `n → n/2`                    | `T(n)=T(n/2)+O(1)` | **O(log n)**    |

Your function falls into the **second** case (`n → n-2`), so:

* **Time Complexity:** **O(n)**
* **Auxiliary Space:** **O(n)** (due to recursion stack)

The key distinction is:

* **Subtracting a constant** (`n-2`) → **linear** time.
* **Dividing by a constant** (`n/2`) → **logarithmic** time.


## Reverse using single variable


````python
def reverse(arr,i):
    n = len(arr)
    if i >=n//2:
        return
    arr[i],arr[n-i-1] = arr[n-i-1],arr[i]
    reverse(arr,i+1)
````

````python
reverse(l,0)
l
````

## Palindrome

````python
def palindrome(word,l,h):
    if l>=h:
        return True
    if word[l]!=word[h]:
        return False
    return palindrome(word,l+1,h-1)

print(palindrome("abccb",0,4))
````

## Palindrome with single variable

````python
def palindrome(word,i):
    n = len(word)
    if i > (n//2):
        return True
    if word[i] != word[n-i-1]:
        return False
    return palindrome(word,i+1)

print(palindrome("abcba",0))
````
