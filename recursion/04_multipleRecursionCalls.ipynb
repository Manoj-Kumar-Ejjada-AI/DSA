#### In this type of recursion multiple calls of a funcion happens inside a function

The Fibonacci series is a sequence of numbers where each number is the sum of the two preceding ones, starting from 0 and 1.

The sequence begins as: 0, 1, 1, 2, 3, 5, 8, 13, 21, ...

F(0) = 0,
F(1) = 1,
F(2) = 1

lets write a normal function that gives the fibonacci number (number in fibonacci series) for a number `N`

```python
def fibonacci_number(n):
  if n<=1:
    return n
  a,b = 0,1
  for i in range(n):
    a,b = b, a+b
    return a

print(fibonacci_number(2))
```

Below is the code for fibonacci series

```python
def fibonacci_series(n):
  a = 0
  b = 1
  for i in range(n):
    print(a, end = " ")
    a,b = b, a+b

fibonacci_series(3)
```

Now, lets try this with recursion with multiple calls

```python
def fibonacci(n):
    if n <=1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
print(fibonacci(2))
```

The above code only prints a fibonacci number for a given `n`, in order to get fibonacci series we can just add a for loop in print statement using list comprehension as folows:

```python
def fibonacci_series(n):
  if n<=1:
    return n
  return fibonacci_series(n-1) + fibonacci_series(n-2)
print([fibonacci_series(i) for i in range(6)])
```
