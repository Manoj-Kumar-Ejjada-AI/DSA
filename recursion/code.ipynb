## Print a name `N` times

```python
def print_name_n_times(i,n):
  if i>n:
    return
  print("Name")
  print_name_n_times(i+1,n)
```

```python
print_name_n_times(1,3)
```

## Print from 1 to `N`

```python
def print_1_to_n(i,n):
  if i>n:
    return
  print(i)
  print_1_to_n(i+1,n)
```

```python
print_1_to_n(1,3)
```

## Print Linearly from 1 to `N`

```python
def print_1_to_n_linearly(i,n):
  if i > n:
    return
  print(i, end = " ")
  print_1_to_n_linearly(i+1,n)
```

```python
print_1_to_n_linearly(1,5)
```

## Print Linearly from `N` to 1

```python
def print_n_to_1_linearly(n):
  if n == 0:
    return
  print(n, end = " ")
  print_n_to_1_linearly(n-1)
```

```python
print_n_to_1_linearly(5)
```

## Print 1 to `N` by backtracking

```python
def print_1_to_n_by_bcktrk(n):
  if n < 1:
    return
  print_1_to_n_by_bcktrk(n-1)
  print(n)
```

```python
print_1_to_n_by_bcktrk(5)
```

## Print `N` to 1 by backtracking

```python
def print_n_to_1_by_bcktrk(i,n):
  if i > n:
    return
  print_n_to_1_by_bcktrk(i+1,n)
  print(i)
```

```python
print_n_to_1_by_bcktrk(1,5)
```
