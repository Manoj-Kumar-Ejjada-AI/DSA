# Parameterized Recursion
## Print sum of first `N` numbers 

```python
def _parameterized_sum(n,sum):
    if n < 1:
        print(sum)
        return
    _parameterized_sum(n-1,sum+n)
_parameterized_sum(3,0)
```

#### Time Comlexity O(N)
#### Space Complexity O(N)

### Another way by incrementing i

```python
def _parameterized_sum(i,n, sum):
    if i > n:
        print(sum)
        return
    _parameterized_sum(i+1,n,sum+i)
_parameterized_sum(1,3,0)
```

#### Time Comlexity O(N)
#### Space Complexity O(N)

## Factorial

```python
def factorial(n,fact):
    if n<1:
        print(fact)
        return
    factorial(n-1,fact*n)
factorial(5,1)
    
```

#### Time Comlexity O(N)
#### Space Complexity O(N)
