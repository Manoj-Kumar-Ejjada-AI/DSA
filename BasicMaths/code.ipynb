## Count digits in a number

### Brute Force

#### Time complexity O(n)
#### Space complexity O(1)

```python
def countDigits(n):
    digits = 0
    while n>0:
        n = n//10
        digits+=1

    return digits
print(countDigits(1))
```

### Optimal Solution

#### Time complexity O(1)
#### Space complexity O(1)

```python
import math
def countDigits(n):
    return int(math.log10(n))+1
print(countDigits(12345))
```

### Reverse a Number


#### Time Complexity O(log₁₀N)
The number of iterations in the loop depends on the number of digits in the number N. Since each digit is processed once using constant time operations (modulo, division, multiplication, and addition), the total time taken is proportional to the number of digits, which is log₁₀N.
#### Space complexity O(1)

```python
def reverseNumber(n):
    rev = 0
    while n>0:
        digit = n%10
        rev = rev*10 + digit
        n = n//10
    return rev
print(reverseNumber(123450))
```

### Palindrome Number

#### Time complexity O(log₁₀N)
#### Space complexity O(1)

```python
def palindromeNumber(n):
    Num = n
    rev = 0
    while n>0:
        digit = n%10
        rev = rev*10 + digit
        n = n//10

    if Num == rev:
        return True
    return False

print(palindromeNumber(12321))
```

### GCD of Two Numbers


#### Brute Force

#### Time complexity O(min(N1,N2))
#### Space complexxity O(1)

```python
def gcd(num1,num2):
    for i in range(min(num1,num2),0,-1):
        if num2%i == 0 and num1%i ==0:
            return i
print(gcd(9,12))
```

#### Optimal

#### Time complexity O(log(Min(N1,N2)))
#### Space complexxity O(1)

```python
def gcd(num1,num2):
    while num1>0 and num2>0:
        if num1>num2:
            num1 = num1 % num2
        else:
            num2 = num2 % num1
    if num2 == 0:
        return num1
    return num2
print(gcd(90,10))
```

### Armstrong Number

#### Brute Force

#### Time complexity O(logN+1) where N is the input number. The time complexity is determined by the number of digits in the input integer N. In the worst case when N is a multiple of 10 the number of digits in N is log10 N + 1.
#### Space complexxity O(1)

```python
import math
class solution:

    def isArmstrong(self,Num):
        digits = int(math.log10(Num))+1
        OrgNum = Num
        sumOfDigits = 0
        while Num>0:
            rem = Num%10
            sumOfDigits = rem ** digits + sumOfDigits
            Num = Num//10
        if sumOfDigits == OrgNum:
            return True
        return False

checkArmstrong = solution()
print(checkArmstrong.isArmstrong(153))
print(checkArmstrong.isArmstrong(152))


```

### Print all divisors


#### Brute Force

#### Time complexity O(N)
#### Space complexxity O(N)

```python
class solution:
    def printAllDivisors(Num):
        divisors = []
        for i in range(1, Num):
            if Num%i == 0:
                divisors.append(i)
        return divisors

print(solution.printAllDivisors(15))
```

#### Optimal

#### Time complexity O(sqrt(N))
#### Space complexxity O(N)

```python
class solution:
    def printAllDivisors(Num):
        divisors = []
        for i in range(1, int(Num**0.5)+1):
            if Num%i == 0:
                divisors.append(i)
                if i != Num //i:
                    divisors.append(Num//i)

        return divisors
print(solution.printAllDivisors(15))

```

### Prime or not

#### Brute Force

#### Time complexity O(n)
#### Space complexxity O(1)

```python
class solution:
    def isPrime(self, Num):
        for i in range(2,Num):
            if Num%i == 0:
                return False
        return True
print(solution().isPrime(13))
```

#### Optimal

#### Time complexity O(sqrt(N))
#### Space complexxity O(1)

```python
class solution:
    def isPrime(self,Num):
        for i in range(2, int(math.sqrt(Num))+1):
            if Num % i == 0:
                return False
        return True
print(solution().isPrime(2))
```
