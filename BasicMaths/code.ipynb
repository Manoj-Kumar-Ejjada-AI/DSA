## Count digits in a number

### Brute Force

#### Time complexity O(n)
#### Space complexity O(1)

```python
def countDigits(n):
    digits = 0
    while n>0:
        n = n//10
        digits+=1

    return digits
print(countDigits(1))
```

### Optimal Solution

#### Time complexity O(1)
#### Space complexity O(1)

```python
import math
def countDigits(n):
    return int(math.log10(n))+1
print(countDigits(12345))
```

### Reverse a Number


#### Time Complexity O(n)
#### Space complexity O(1)

```python
def reverseNumber(n):
    rev = 0
    while n>0:
        digit = n%10
        rev = rev*10 + digit
        n = n//10
    return rev
print(reverseNumber(123450))
```
