## Swap Two Numbers
<a id="SwapTwoNumbers"></a>
#### `XOR` of two same numbers is 0, and `XOR` of 0 and any Number `N` is `N`  

```python
def SwapTwoNumbers(Num1,Num2):
    Num1 = Num1 ^ Num2
    Num2 = Num1 ^ Num2
    Num1 = Num1 ^ Num2
    return Num1, Num2

print(SwapTwoNumbers(5,4))
```

## Check if ith bit is set or not
<a id = "check-if-ith-bit-is-set-or-not"></a>

### To check if ith bit is set or not there are two ways

## With left shift opertator
<a id = "left-shift"></a>

```python
def check_ith_bit(Num, i):
    if Num & 1<<i >0:
        return True
    return False
print(check_ith_bit(5,1))  # 5 = 1 0 1, 1 << 1 = 0 1 0
```

### With Right Shift Operator
<a id = "right-shift"></a>

```python
def check_ith_bit_with_right_shift(Num, i):
    if Num>>i & 1 >0:
        return True
    return False
print(check_ith_bit_with_right_shift(5,2))
```

## Set ith Bit
<a id = "set-ith-bit"></a>

```python
def set_ith_bit(Num,i):
  return Num |1<<i

print(set_ith_bit(9,1))
```

### Clear ith bit

<a id = "clear-ith-bit"></a>

```python
def clear_ith_bit(Num,i):
  return Num ^ 1 <<i

print(clear_ith_bit(11,1))
```

### Toggle ith bit

<a id = "toggle-ith-bit"></a>

```python
def toggle_ith_bit(Num,i):
  return Num & ~(1 <<i)

print(toggle_ith_bit(11,1))
```

### Remove last set bit

<a id = "remove-last-set-bit"></a>

```python
def remove_last_set_bit(Num):
  return Num & Num-1

print(remove_last_set_bit(15))  # 15 = 1111 , 15-1 = 14 = 1110 to remove last set bit, 1111 & 1110
print(remove_last_set_bit(10))  # 10 = 1010 , 10-1 = 9 = 1001 to remove last set bit, 1010 & 1001
```

### Count set bits

<a id = "count-set-bits"></a>

```python
# Brute Force
def count_set_bits(Num):
  count = 0
  while Num>0:
    if Num%2!=0:
      count+=1
    Num = Num//2
  return count
print(count_set_bits(27))
```

```python
# With Bit Manipulation better performance
def count_set_bits(Num):
  count = 0
  while Num>0:
    count+=Num & 1
    Num = Num>>1
  return count
print(count_set_bits(27))
```

```python
def count_set_bits(Num):
  count = 0
  while Num !=0:
    Num = Num&Num-1
    count+=1
  return count
print(count_set_bits(27))
```

### Check if number is power of 2 or not

<a id = "power-of-2"></a>

```python
def power_of_2(Num):
  if Num & Num-1 == 0:
    return True
  return False

print(power_of_2(6))
print(power_of_2(8))
```

### Power set
<a id = "power-set"></a>

```python
def power_set(arr):
  n = len(arr)
  ans = []
  for num in range(1<<n):
    _list = []
    for i in range(n):
      if 1 & num >> i != 0:
        _list.append(arr[i])
    ans.append(_list)
  return ans
print(power_set([1,2,3]))
```

### Single Number I

<a id = "single-number-1"></a>

```python
def single_number_1(arr):
  dic = {}
  for num in arr:
    if num not in dic:
      dic[num] = 1
    else:
      dic[num]+=1
  for num in dic:
    if dic[num] == 1:
      return num
print(single_number_1([3,3,1,1,5,5,2,2,4,4,3,3,7]))
```

```python
def single_number_1(arr):
  xor = 0
  for num in arr:
    xor = xor ^ num
  return xor
print(single_number_1([3,3,1,1,5,5,2,2,4,4,3,3,7]))
```

### Single Number II

<a id = "single-number-2"></a>

```python
def single_number_2(arr):
  res = 0
  for bit in range(31):
    count_ones = 0
    for num in arr:
      if (num>>bit) & 1 ==1:
        count_ones+=1
    if count_ones%3>0:
      res+= 2**bit
  return res
print(single_number_2([3,3,3,4,4,4,5,6,6,6]))
```

```python
def single_number_2(arr):
  num = len(arr)

  for i in range(1,len(arr),3):
    if arr[i] != arr[i-1] or arr[i] != arr[i+1]:
      return arr[i-1]
  return arr[num-1]

print(single_number_2([3,2,2,2]))
```
