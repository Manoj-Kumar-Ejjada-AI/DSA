## Swap Two Numbers
<a id="SwapTwoNumbers"></a>
#### `XOR` of two same numbers is 0, and `XOR` of 0 and any Number `N` is `N`  

```python
def SwapTwoNumbers(Num1,Num2):
    Num1 = Num1 ^ Num2
    Num2 = Num1 ^ Num2
    Num1 = Num1 ^ Num2
    return Num1, Num2

print(SwapTwoNumbers(5,4))
```

## Check if ith bit is set or not
<a id = "check-if-ith-bit-is-set-or-not"></a>

### To check if ith bit is set or not there are two ways

## With left shift opertator
<a id = "left-shift"></a>

```python
def check_ith_bit(Num, i):
    if Num & 1<<i >0:
        return True
    return False
print(check_ith_bit(5,1))  # 5 = 1 0 1, 1 << 1 = 0 1 0
```

### With Right Shift Operator
<a id = "right-shift"></a>

```python
def check_ith_bit_with_right_shift(Num, i):
    if Num>>i & 1 >0:
        return True
    return False
print(check_ith_bit_with_right_shift(5,2))
```

## Set ith Bit
<a id = "set-ith-bit"></a>

```python
def set_ith_bit(Num,i):
  return Num |1<<i

print(set_ith_bit(9,1))
```

### Clear ith bit

<a id = "clear-ith-bit"></a>

```python
def clear_ith_bit(Num,i):
  return Num ^ 1 <<i

print(clear_ith_bit(11,1))
```

### Toggle ith bit

<a id = "toggle-ith-bit"></a>

```python
def toggle_ith_bit(Num,i):
  return Num & ~(1 <<i)

print(toggle_ith_bit(11,1))
```

### Remove last set bit

<a id = "remove-last-set-bit"></a>

```python
def remove_last_set_bit(Num):
  return Num & Num-1

print(remove_last_set_bit(15))  # 15 = 1111 , 15-1 = 14 = 1110 to remove last set bit, 1111 & 1110
print(remove_last_set_bit(10))  # 10 = 1010 , 10-1 = 9 = 1001 to remove last set bit, 1010 & 1001
```

### Count set bits

<a id = "count-set-bits"></a>

```python
# Brute Force
def count_set_bits(Num):
  count = 0
  while Num>0:
    if Num%2!=0:
      count+=1
    Num = Num//2
  return count
print(count_set_bits(27))
```

```python
# With Bit Manipulation better performance
def count_set_bits(Num):
  count = 0
  while Num>0:
    count+=Num & 1
    Num = Num>>1
  return count
print(count_set_bits(27))
```

```python
def count_set_bits(Num):
  count = 0
  while Num !=0:
    Num = Num&Num-1
    count+=1
  return count
print(count_set_bits(27))
```

### Check if number is power of 2 or not

<a id = "power-of-2"></a>

```python
def power_of_2(Num):
  if Num & Num-1 == 0:
    return True
  return False

print(power_of_2(6))
print(power_of_2(8))
```

### Power set
<a id = "power-set"></a>

```python
def power_set(arr):
  n = len(arr)
  ans = []
  for num in range(1<<n):
    _list = []
    for i in range(n):
      if 1 & num >> i != 0:
        _list.append(arr[i])
    ans.append(_list)
  return ans
print(power_set([1,2,3]))
```

### Single Number I

<a id = "single-number-1"></a>

```python
def single_number_1(arr):
  dic = {}
  for num in arr:
    if num not in dic:
      dic[num] = 1
    else:
      dic[num]+=1
  for num in dic:
    if dic[num] == 1:
      return num
print(single_number_1([3,3,1,1,5,5,2,2,4,4,3,3,7]))
```

```python
def single_number_1(arr):
  xor = 0
  for num in arr:
    xor = xor ^ num
  return xor
print(single_number_1([3,3,1,1,5,5,2,2,4,4,3,3,7]))
```

### Single Number II

<a id = "single-number-2"></a>

```python
def single_number_2(arr):
  res = 0
  for bit in range(31):
    count_ones = 0
    for num in arr:
      if (num>>bit) & 1 ==1:
        count_ones+=1
    if count_ones%3>0:
      res = res | 1 << bit
  return res
print(single_number_2([3,4,3,4,3,4,5,6,7,7,6,7,6]))
```

```python
def single_number_2(arr):
  num = len(arr)
  arr.sort()
  for i in range(1,len(arr),3):
    if arr[i] != arr[i-1]:
      return arr[i-1]
  return arr[num-1]

print(single_number_2([3,4,3,4,3,4,5,6,7,7,6,7,6]))
```

```python
def single_number_2(arr):
  ones = 0
  twos = 0
  for i in arr:
    ones = (ones^i)&(~twos)
    twos = (twos^i)&(~ones)
  return ones
print(single_number_2([3,4,3,4,3,4,5,6,7,7,6,7,6]))
```

### Single Number III

<a id = "single-number-3"></a>

```python
def single_number_3(arr):
  xor = 0
  first = 0
  second = 0
  # Find XOR of all numbers
  for i in arr:
    xor^=i
  # as xor resulted in bits which are different in both unique numbers
  # lets take right most bit which is different and bucket the numbers in array into two buckets based on this bit
  # if this bit is 1 it goes to 1 bucket, if 0 then diffferent bucket
  xor = (xor & xor-1) ^ xor
  for num in arr:
    if num & xor>0:
      first^=num
    else:
      second^=num
  return first, second
print(single_number_3([1,2,3,4,2,1,3,4,5,6]))
```

### XOR of Numbers in given range

<a id = "xor-in-given-range"></a>

```python
def xor_of_range(num):
  if num%4 == 1:
    return 1
  elif num%4 == 2:
    return num+1
  elif num%4 == 3:
    return 0
  return num
print(xor_of_range(10))
```

```python
def xor_of_range(num):
  if num&3 == 1:
    return 1
  elif num&3 == 2:
    return num+1
  elif num&3 == 3:
    return 0
  return num
print(xor_of_range(10))
```

```python
def xor_in_custom_range(start,end):
  # XOR in range 5 to 9 = (1^2^3^4) ^ (1^2^3^4^5^6^7^8^9) as 1,2,3,4 cancel out
  return xor_of_range(start-1)^xor_of_range(end)
print(xor_in_custom_range(4,9))
```

### didvide two numbers

<a id = "divide-two-numbers"></a>


```python
def divide_two_numbers(num,den):
  sign = 1
  if num == den:
    return 1
  if (num >= 0 and den <0) | (num<0 and den>0):
    sign = -1
  res = 0
  num, den = abs(num), abs(den)
  while num>= den:
    count = 0
    while num >= den<<(count+1):
      count+=1
    res+=1<<count
    num -= den<<count
  return res*sign

print(divide_two_numbers(-29,3))
```
