```python
class solution:
    def searchNumInSortedArray(self,arr, Num):
        l = 0
        h = len(arr)-1
        while l <= h:
            mid = (l+h)>>1
            if Num == arr[mid]:
                return mid
            elif Num > arr[mid]:
                l = mid+1
            elif Num < arr[mid]:
                h = mid-1
        return -1

sol = solution()
sol.searchNumInSortedArray([1,2,3,4,5],2)
```

```python
class solution:
    def searchNumInSortedArray(self, arr,Num, low, high):
        if low>high:
            return -1
        mid = (low + high)>>1
        if arr[mid] == Num:
            return mid
        elif arr[mid]< Num:
            return self.searchNumInSortedArray(arr, Num, mid+1, high)
        else:
            return self.searchNumInSortedArray(arr,Num,  low, mid-1)

    def search(self, arr, Num):
        return self.searchNumInSortedArray( arr, Num, 0, len(arr)-1)
sol = solution()
print(sol.search([1,2,3,4,5],4))
```
